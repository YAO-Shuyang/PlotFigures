```python
if exists(join(figdata, f" {code_id}.pkl")) == False:
    PCProp = {
        "MiceID": [],
        "Session": [],
        "Proportion": []
    }
    maze_type = 1
    for mouse in [10212, 10224, 10227, 10232, 10275, 10276]:
        file_idx = np.where(f2['MiceID'] == mouse)[0]
        for s, idx in tqdm(enumerate(file_idx)):
            with open(f2['Trace File'][idx], 'rb') as f:
                trace = pickle.load(f)
                
            is_placecell = np.vstack([
                trace[f'node {i}']['is_placecell'] for i in range(10)
            ])
            is_pc_atleast_oneroute = np.where(np.sum(is_placecell, axis=0) > 0, 1, 0)
            
            PCProp['MiceID'].append(mouse)
            PCProp['Session'].append(s)
            PCProp['Proportion'].append(np.mean(is_pc_atleast_oneroute))
    
    for key in PCProp.keys():
        PCProp[key] = np.array(PCProp[key])
        
    with open(join(figdata, f"{code_id}.pkl"), 'wb') as f:
        pickle.dump(PCProp, f)
        
    PCPropDF = pd.DataFrame(PCProp)
    #excel
    PCPropDF.to_excel(join(figdata, f"{code_id}.xlsx"), index=False)
    print(len(PCPropDF))
else:
    with open(join(figdata, f"{code_id}.pkl"), 'rb') as f:
        PCProp = pickle.load(f)
        
print_estimator(PCProp["Proportion"])
```
